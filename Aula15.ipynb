```python
notas = []
for i in range(1, 11):
  aluno = input(f"Digite o nome do aluno {i}°: ")

  nota1 = float(input("Digite a primeira nota: "))
  nota2 = float(input("Digite a segunda nota: "))
  media_alunos = (nota1 + nota2) / 2
  notas.append(media_alunos)

media_turma = sum(notas) / len(notas)
print(f"media turma: {media_turma}")
```

```python
#Exercicio 09 - Registro em Python (Desafio Prático)
from dataclasses import dataclass
@dataclass
class livro:
  titulo: str
  autor: str
  ano: int

livros = []

for i in range(2):
  print(f"\n--- Cadastro do Livro {i+1} ---")
  nome_li = input(f"Digite o nome do {i+1}° livro: ")
  autor_name = input(f"Digite o nome do {i+1}° autor: ")
  ano = int(input(f"Digite o ano de publicação do {i+1}° livro: "))

  livro_novo = (livro(titulo=nome_li, autor=autor_name, ano=ano))
  livros.append(livro_novo)

mais_antigo = livros[0]

for l in livros:
  if l.ano < mais_antigo.ano:
    mais_antigo = l

print("\n----------RESULTADO-------------")
print(f"nome do livro: {mais_antigo.titulo}")
print(f"ano publicação: {mais_antigo.ano}")
print("--------------------------------")
```
